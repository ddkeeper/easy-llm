```python
# 1. 变量与基础类型
num = 10          # 整数
arr = [1,2,3,4,5] # 其他语言中叫数组（array），python 中叫列表（list）

# 2. 函数定义
def add(x, y):
    return x + y


res = add(arr[0], arr[1])
print(f"arr[0] 值: {arr[0]}")
print(f"arr[1] 值: {arr[1]}")
print(f"add(arr[0], arr[1]) 的结果：{add(arr[0], arr[1])}")
```

```python
# 3. 分支 + 循环
n = len(arr) # 取数组的长度
for i in range(n): # 0,1,2,...,n-1
    if arr[i] > 3: 
        print(f"第{i}个位置: {arr[i]}")

i = 0
while i < n:
    if arr[i] > 3:
        print(f"第{i}个位置: {arr[i]}")
    i += 1
```

```python
# 4. 列表分片（左闭右开）
sub_arr = arr[1:4] # 得到 [2,3,4]，下标范围为 [1, 4)

print(f"原始数组 arr：{arr}")
print(f"arr[1:4]：{sub_arr}")
```

```python
# 5. 一个简单的类；简单来说，类 = 属性 + 方法（类内部的函数）
class Module:
    def __init__(self, dim):
        # __init__ 是构造函数，创建实例时自动执行，用于初始化属性
        self.dim = dim  # 实例属性
    
    # 定义一个方法
    def forward(self, x):
        return x * self.dim # 通过 self.xxx 访问实例的属性和方法

# 初始化实例并使用
net = Module(dim=4)
print(f"net.dim: {net.dim}") # 通过 self.xxx 访问实例
print(f"net.forward(3) 的结果: {net.forward(3)}")  # 通过 self.xxx() 访问方法
```

```python
import torch
import torch.nn as nn

# 1. 创建张量 Tensor（深度学习的基础数据结构，本质上是多维数组，但支持梯度自动计算,以及搬到 GPU 上进行并行计算）
# # 一维常量张量
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True) 
print(f"张量 x 的值：{x}")
print(f"张量 x 的形状：{x.shape}")  
print(f"张量 x 是否开启梯度追踪：{x.requires_grad}")

# 二维常量张量
y = torch.tensor([[1.0, 2.0, 3.0], [1.0, 2.0, 3.0], [1.0, 2.0, 3.0]]) # 二维张量
print(f"\n张量 y 的形状：{y.shape}") 
print(f"张量 y 是否开启梯度追踪：{y.requires_grad}")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# 将张量迁移到目标设备
y_gpu = y.to(device)
print(f"迁移后张量 y_gpu 所在的设备：{y_gpu.device}")
```

```python
# 2. 矩阵乘法（Transformer 大模型最核心的基础运算）
w = torch.ones(3, 3)
y = x @ w  # 写法等价于 torch.matmul(x, w)
print(f"矩阵 x：{x}")
print(f"\n矩阵 w：{w}")
print(f"\n矩阵乘法结果 y：{y}")
# (3) * (3, 3) = (3)
```

```python
# 3. 一个简单网络的定义（PyTorch 中所有网络都继承 nn.Module 基类，基类有的属性和方法，子类都会有）
class Net(nn.Module):
    def __init__(self):
        super().__init__()  # 固定写法，必须调用父类的构造函数
        self.fc = nn.Linear(10, 2)  # 全连接层：输入维度 10，输出维度 2
    
    # 前向传播函数：定义数据在网络里的计算流程
    def forward(self, x):
        return self.fc(x) # x * W_fc, (3, 10) * (10, 2) -> (3, 2)

# 初始化网络实例并运行
net = Net()
input_x = torch.ones(3, 10)  # (3, 10)
output = net(input_x)  # 把实例名 net 当作函数来用，相当于 net.forward(input_x)，这就是 “继承自 nn.Module 类” 的魅力
print(f"\n网络输入形状：{input_x.shape}")
print(f"网络输出形状：{output.shape}")
```
